```python
import pandas as pd
```

```python
df = pd.DataFrame(
    {
        "Point":['P1','P2','P3','P4','P5','P6','P7','P8','P9','P10','P11','P12'],
        "X":[3,4,5,6,7,6,7,8,3,2,3,2],
        "Y":[7,6,5,4,3,2,2,4,3,6,5,4]
    }
)
```

```python
df
```

```python
from sklearn.cluster import DBSCAN
```

```python
eps = 1.9
minPts = 4
```

```python
clustering = DBSCAN(eps=eps,min_samples=minPts)
```

```python
val = df[['X','Y']]
```

```python
clus = clustering.fit_predict(val)
```

```python
df['cluster'] = clus
```

```python
df
```

```python
df0 = df[df['cluster']==0]
df1 = df[df['cluster']==1]
df_noise = df[df['cluster']==-1]
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df0.values[:,1],df0.values[:,2],color='blue')
plt.scatter(df1.values[:,1],df1.values[:,2],color='green')
plt.scatter(df_noise.values[:,1],df_noise.values[:,2],color='red')
```
